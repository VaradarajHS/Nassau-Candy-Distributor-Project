```python
product_analysis = (
    df.groupby(['Division', 'Product Name'])
    .agg(
        Total_Sales=('Sales', 'sum'),
        Total_Cost=('Cost', 'sum'),
        Total_Profit=('Gross Profit', 'sum'),
        Total_Units=('Units', 'sum')
    )
    .reset_index()
)
```

```python
product_analysis['Gross Margin%']=(
    product_analysis['Total_Profit'] /
    product_analysis['Total_Sales']
)*100
product_analysis['Profit per Unit'] = (
    product_analysis['Total_Profit'] /
    product_analysis['Total_Units']
)
product_analysis['Revenue Contribution %'] = (
    product_analysis['Total_Sales'] /
    product_analysis['Total_Sales'].sum()
) * 100
product_analysis['Profit Contribution %'] = (
    product_analysis['Total_Profit'] /
    product_analysis['Total_Profit'].sum()
) * 100
```

```python
pareto_products = product_analysis[
    ['Division', 'Product Name', 'Total_Sales', 'Total_Profit']
    ].copy()
```

```python
pareto_products = pareto_products.sort_values(
    'Total_Sales',
    ascending=False
)
```

```python
pareto_products['Revenue Contribution%']=(
    pareto_products['Total_Sales'].cumsum()
    / pareto_products['Total_Sales'].sum()
) * 100
```

```python
pareto_products['Profit Contribution%']=(
    pareto_products['Total_Profit'].cumsum()
    / pareto_products['Total_Profit'].sum()
) * 100
```

```python
revenue_80 = pareto_products[
    pareto_products['Revenue Contribution%'] <= 80
    ]
```

```python
profit_80 = pareto_products[
pareto_products['Profit Contribution%'] <= 80
]

print("Products contributing to 80% of Revenue:")
print(len(revenue_80))

print("\nProducts contributing to 80% of Profit:")
print(len(profit_80))

print("\nTop Revenue-Contributing Products:")
display(
    pareto_products[
        ['Division', 'Product Name', 'Total_Sales',
         'Revenue Contribution%', 'Profit Contribution%']
    ].head(20)
)
```

```python
print('Products contributing to 80% of Revenue:')
display(
    revenue_80[
        ['Division', 'Product Name',
         'Total_Sales',
         'Revenue Contribution%',
         'Profit Contribution%']
    ]
)

print('\nProducts contributing to 80% of Profit:')
display(
    profit_80[
        ['Division', 'Product Name',
         'Total_Sales',
         'Total_Profit',
         'Revenue Contribution%',
         'Profit Contribution%']
    ]
)
```

```python
revenue_division_concentration=(
    revenue_80
    .groupby('Division')
    .agg(
        Products=('Product Name', 'count'),
        Total_Sales=('Total_Sales', 'sum'),
        Total_Profit=('Total_Profit', 'sum')
    )
    .reset_index()
)
revenue_division_concentration['Revenue%']=(
    revenue_division_concentration['Total_Sales']
    / revenue_division_concentration['Total_Sales'].sum()
) * 100
revenue_division_concentration['Profit%']=(
    revenue_division_concentration['Total_Profit']
    / revenue_division_concentration['Total_Profit'].sum()
) * 100
revenue_division_concentration.sort_values(
    'Total_Sales',
    ascending=False
)
```

```python
cost_analysis = product_analysis[
    ['Division', 'Product Name', 'Total_Sales',
     'Total_Cost', 'Total_Profit', 'Gross Margin%']
].copy()

cost_analysis['Cost % of Sales'] = (
    cost_analysis['Total_Cost'] /
    cost_analysis['Total_Sales']
) * 100

cost_analysis = cost_analysis.sort_values(
    'Cost % of Sales',
    ascending=False
)
```

```python
display(cost_analysis.head(10))
```

```python
cost_heavy_products = cost_analysis[
    (cost_analysis['Total_Cost'] / cost_analysis['Total_Sales'] * 100 > 50) &
    (cost_analysis['Gross Margin%'] < 50)
].copy()

cost_heavy_products['Cost % of Sales'] = (
    cost_heavy_products['Total_Cost'] /
    cost_heavy_products['Total_Sales'] * 100
)

cost_heavy_products = cost_heavy_products.sort_values(
    'Cost % of Sales',
    ascending=False
)
```

```python
display(cost_heavy_products)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(
    cost_analysis['Total_Sales'],
    cost_analysis['Total_Cost']
)
plt.xlabel('Total Sales')
plt.ylabel('Total Cost')
plt.title('Cost vs Sales Analysis')
plt.show()
```

```python
product_recommendations = cost_analysis.copy()
product_recommendations['Recommendation'] = 'Monitor'
product_recommendations.loc[
    (product_recommendations['Cost % of Sales']> 80) &
    (product_recommendations['Gross Margin%']< 30),
    'Recommendation'
    ]='Discontinuation Review'
product_recommendations.loc[
    (product_recommendations['Cost % of Sales']> 50) &
    (product_recommendations['Gross Margin%']< 50) &
    (product_recommendations['Recommendation'] == 'Monitor'),
    'Recommendation'
    ]= 'Cost Renegotiation'
product_recommendations.loc[
    (product_recommendations['Gross Margin%']> 50) &
    (product_recommendations['Recommendation'] == 'Monitor'),
    'Recommendation'
    ]='Repricing'
display(
    product_recommendations[
        ['Division', 'Product Name', 'Total_Sales',
         'Total_Cost', 'Gross Margin%',
         'Cost % of Sales', 'Recommendation']
    ].sort_values('Cost % of Sales', ascending=False)
)
```
